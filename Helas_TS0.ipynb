# Introducción
En este trabajo graficamos una señal senoidal a la cual le vamos cambiando la frecuencia para asi analizar que pasa en cietos puntos de interes







    

```python
import numpy as np
import matplotlib.pyplot as plt

fs = 1000 #Cantidad de frecuencia fija
N = 1000 #Cantidad de muestras fijas

#Variables para ir cambiando y que cambie mi funcion
vmax = 3 #Amplitud maxima de mi señal
dc = 0 #Desplazamiento, me mueve mi señal que queda fija com vmax = 2 para el centro = dc
ff = 3 #Frecuencia variable
ph = 0 #La fase

def mi_funcion_sen( vmax = 1, dc = 0, ff = 1, ph=0, nn = N, fs = fs):
    
    xx = (vmax * np.sin(2 * np.pi * ff * np.arange(nn)/fs + ph)) +dc
    tt = np.arange(N)/fs
    return (tt,xx)

tt, xx = mi_funcion_sen( vmax = vmax, dc = dc, ff = ff, ph = ph, nn = N, fs = fs)

plt.plot(tt, xx)
plt.title("Señal senoidal")
plt.xlabel("Tiempo [t]")
plt.ylabel("Dc [v]")
```

# 1) Que pasa en la frecuencia de Nyquist-Shannon (500hz)
La frecuencia de Nyquist-Shannon es justo la mitad de la cantidad de muestras que tenas de la señal y es el ultimo limite en el cual es teoricamente posible reconstruir la señal si se dispone de un pasa bajo perfecto, en la practica sin embargo es imposible reconstruir una señal apartir de una frecuencia que sea la mitad de tu frecuencia de muestreo ya que el grafico se ve de la siguiente forma:


```python
import numpy as np
import matplotlib.pyplot as plt

fs = 1000 #Cantidad de frecuencia fija
N = 1000 #Cantidad de muestras fijas

#Variables para ir cambiando y que cambie mi funcion
vmax = 3 #Amplitud maxima de mi señal
dc = 0 #Desplazamiento, me mueve mi señal que queda fija com vmax = 2 para el centro = dc
ff = 500 #Frecuencia variable
ph = 0 #La fase

def mi_funcion_sen( vmax = 1, dc = 0, ff = 1, ph=0, nn = N, fs = fs):
    
    xx = (vmax * np.sin(2 * np.pi * ff * np.arange(nn)/fs + ph)) +dc
    tt = np.arange(N)/fs
    return (tt,xx)

tt, xx = mi_funcion_sen( vmax = vmax, dc = dc, ff = ff, ph = ph, nn = N, fs = fs)

plt.plot(tt, xx)
plt.title("Señal con frecuencia 500hz")
plt.xlabel("Tiempo [t]")
plt.ylabel("Dc [v]")
```

# 2) ¿Que pasa a 999hz?
Como se puede ver con una frecuencia de 999hz, se ve como una señal de -1hz, esto ocurre debido a un fenomeno llamado aliasing que ocurre en uns señal reconstruida submuestreada es decir donde la frecuencia de muestreo es menor al doble de de la frecuenia de la señal en este caso la frecuencia de muestreo es 1000hz y la de la señal 999hz.
Se ve como una señal de -1hz debido a la periodicidad de espectros que aparece por discretizar la señal en un principio ya que al discretizar el tiempo en muestras, volves periodica a la señan entonces cuando la computadora recupera la señal debido al aliasing y no tener datos fidelignos, busca el centro de la replica mas cercana que en este caso ocurre en 1000hz y hace la cuenta 999-1000 para asi reconstruir la señal como si fuese de -1hz 

```python
import numpy as np
import matplotlib.pyplot as plt

fs = 1000 #Cantidad de frecuencia fija
N = 1000 #Cantidad de muestras fijas

#Variables para ir cambiando y que cambie mi funcion
vmax = 3 #Amplitud maxima de mi señal
dc = 0 #Desplazamiento, me mueve mi señal que queda fija com vmax = 2 para el centro = dc
ff = 999 #Frecuencia variable
ph = 0 #La fase

def mi_funcion_sen( vmax = 1, dc = 0, ff = 1, ph=0, nn = N, fs = fs):
    
    xx = (vmax * np.sin(2 * np.pi * ff * np.arange(nn)/fs + ph)) +dc
    tt = np.arange(N)/fs
    return (tt,xx)

tt, xx = mi_funcion_sen( vmax = vmax, dc = dc, ff = ff, ph = ph, nn = N, fs = fs)

plt.plot(tt, xx)
plt.title("Señal 999hz")
plt.xlabel("Tiempo [t]")
plt.ylabel("Dc [v]")
```

# 3) ¿Que pasa en 1001hz y 2001hz?
Como se hablo en el punto anterior, al discretizar el tiempo el espectro de la señal de volvio periodico es decir aparecen replicas exactas de la funcion original centradas en multiplos exactos de la frecuencia de muestreo entonces, cada vez que estamos en un multiplo exacto de nuestro fs en este caso cualquier multiplo de 1000 es como estar en 0hz y arrancar de vuelta lo que quiere decir que tanto en 1001hz como en 2001hz se va a ver como si estuviesemos en 1hz

```python
import numpy as np
import matplotlib.pyplot as plt

fs = 1000 #Cantidad de frecuencia fija
N = 1000 #Cantidad de muestras fijas

#Variables para ir cambiando y que cambie mi funcion
vmax = 2 #Amplitud maxima de mi señal
dc = 0 #Desplazamiento, me mueve mi señal que queda fija com vmax = 2 para el centro = dc
ff1 = 1 #Frecuencia variable
ff2 = 1001
ff3 = 2001
ph = 0 #La fase

def mi_funcion_sen( vmax = 1, dc = 0, ff = 1, ph=0, nn = N, fs = fs):
    
    xx = (vmax * np.sin(2 * np.pi * ff * np.arange(nn)/fs + ph)) +dc
    tt = np.arange(N)/fs
    return (tt,xx)

tt, xx = mi_funcion_sen( vmax = vmax, dc = dc, ff = ff1, ph = ph, nn = N, fs = fs)

plt.subplot(3, 1, 1)
plt.plot(tt, xx)
plt.title("Señal 1hz")
plt.xlabel("Tiempo [t]")
plt.ylabel("Dc [v]")

tt, xx = mi_funcion_sen( vmax = vmax, dc = dc, ff = ff2, ph = ph, nn = N, fs = fs)
plt.subplot(3, 1, 2)
plt.plot(tt, xx)
plt.title("Señal 1001hz")
plt.xlabel("Tiempo [t]")
plt.ylabel("Dc [v]")

tt, xx = mi_funcion_sen( vmax = vmax, dc = dc, ff = ff3, ph = ph, nn = N, fs = fs)
plt.subplot(3, 1, 3)
plt.plot(tt, xx)
plt.title("Señal 2001hz")
plt.xlabel("Tiempo [t]")
plt.ylabel("Dc [v]")

plt.tight_layout()
```
